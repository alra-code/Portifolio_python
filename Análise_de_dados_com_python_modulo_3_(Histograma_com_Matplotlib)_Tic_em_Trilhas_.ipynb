##Histograma com Matplotlib

```python
#importações
import pandas as pd
import matplotlib.pyplot as plt
```

```python
#lendo arquivo no drive
df_new=pd.read_excel('/content/drive/MyDrive/CURSOS/Estudos - Data Analitycs/Tic - Análise de dados com paython /Módulo 3/Aula 2/meteorite_landings (1).xlsx')
df_new.head()
```

```python
df_new=df_new[(df_new['reclong']<=180) & (df_new['reclong']>=-180)] #removendo os outiliers
#histograma de latitude
plt.hist(df_new['reclat'])
plt.show()
```

```python
#variando o número bins
plt.hist(df_new['reclat'],bins=5)
plt.show()
```

```python
#histograma de longitude
plt.hist(df_new['reclong'])
plt.show()
```

```python
plt.hist(df_new['reclong'],bins=5, color='green')
plt.show()
```

```python
#histogramas em subplots
plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.hist(df_new['reclong'])
plt.xlabel('Longitude')
plt.ylabel('Frequência')
plt.title('Histograma de Longitude')
plt.subplot(1,2,2)
plt.hist(df_new['reclat'], color='gray')
plt.xlabel('Latitude')
plt.ylabel('Frequência')
plt.title('Histograma de Latitude')
plt.show()
```

```python
#Latitude e lontitude no mesmo gráfico
plt.figure(figsize=(15,5))
plt.hist([df_new['reclat'], df_new['reclong']],bins=50, label=['Latitude','Longitude'])
plt.legend(loc='upper right')
plt.xlabel('Coordenadas')
plt.ylabel('Frequência')
plt.title('Histograma de Latitude e Longitude')
plt.show()
```

```python
plt.figure(figsize=(15,5))
plt.hist([df_new['reclat'], df_new['reclong']],bins=20, label=['Latitude','Longitude']) #alterando a quantidade de barras
plt.legend(loc='upper right')
plt.xlabel('Coordenadas')
plt.ylabel('Frequência')
plt.title('Histograma de Latitude e Longitude')
plt.show()
```
