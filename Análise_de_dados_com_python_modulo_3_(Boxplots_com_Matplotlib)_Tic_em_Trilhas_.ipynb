##Trabalhando com Boxplots

```python
#importações
import pandas as pd
import matplotlib.pyplot as plt
```

```python
#abrindo o dataframe
df=pd.read_excel('/content/sentiment_dataset.xlsx')
df.head()
```

```python
df['Country'].value_counts()
```

```python
df['Sentiment'].values
```

```python
df['Country'].values
```

```python
#removendo espeços extras das strings
df['Country']=df['Country'].str.strip()
df['Sentiment']=df['Sentiment'].str.strip()
```

```python
#filtrando as linhas que correspondem ao pais 'USA'
df_new=df[df['Country']=='USA']
```

```python
#Boxplot
fig,ax=plt.subplots()
ax.boxplot([df_new['Likes']], labels=['USA'])
ax.set_title('País x Likes')
plt.show()
```

```python
#filtrando linhas com sentimentos positivo do país USA
df_new3=df_new[df_new['Sentiment']=='Positive']
df_new['Sentiment'].value_counts()
```

```python
#Subplots de boxplot
fig, ax = plt.subplots()
ax.boxplot([df_new['Likes'], df_new3['Likes']], labels=['USA', 'Positive'])
ax.set_title('País x Likes')
plt.show()
```
