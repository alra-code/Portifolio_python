## Filtrando multiplos valores

```python
regiao_sul= ['Rio Grande do Sul','Santa Catarina','Paraná']
df_populacao_sul=df_populacao[df_populacao['uf'].isin(regiao_sul)]
df_populacao_sul
```

```python
regiao_sul = ['Rio Grande do Sul', 'Santa Catarina', 'Paraná']
df_populacao_sul = df_populacao[(df_populacao['uf'].isin(regiao_sul)) & (df_populacao['populacao'] < 250000)]
df_populacao_sul
```

## Unindo Tabelas

```python
dados_area_urbana=pd.read_csv('/content/drive/MyDrive/CURSOS/Estudos - Data Analitycs/Tic - Análise de dados com paython /Módulo 2/Aula 2/area_urbana.csv')
dados_area_urbana
```

```python
dados_area_urbana.info()
```

```python
dados_area_urbana.head()
```

```python
#unindo os daframes
df_merged = pd.merge(
    left=df_populacao, right=dados_area_urbana,
    how='outer', on=['municipio','uf'],
    suffixes=['_populacao','_area_urbana']
)
```

```python
df_merged.tail()
```

```python
df_merged.info ()
```

```python
df_merged.isna().sum()
```

## Mantendo dados existentes nas duas tabelas

```python
df_merge_inner = pd.merge(
    left=df_populacao, right=dados_area_urbana,
    how='inner', on=['municipio','uf'],
    suffixes=['_populacao','_area_urbana']
)
df_merge_inner.dropna(inplace=True)
```

```python
df_merge_inner.head()
```

```python
df_merge_inner.tail()
```

```python
df_merge_inner.info()
```

## Organizando e Savando o novo dataframe

```python
df_merge_inner['habitante por km²']=df_merge_inner['populacao']/df_merge_inner['area urbana km²']
df_merge_inner
```

```python
df_merge_inner=df_merge_inner.sort_values(by='habitante por km²', ascending=False).reset_index(drop=True)
df_merge_inner
```

```python
df_merge_inner['Ranking de Densidade']=df_merge_inner.index+1
df_merge_inner
```

```python
df_merge_inner=df_merge_inner[
    ['municipio','uf',
     'populacao', 'ranking_populacao',
     'area urbana km²', 'ranking_area_urbana',
     'habitante por km²','Ranking de Densidade']
]
df_merge_inner
```
