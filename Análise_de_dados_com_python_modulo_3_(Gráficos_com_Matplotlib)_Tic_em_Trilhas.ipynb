##Gráficos com Matplolib

```python
#importações
import pandas as pd
import matplotlib.pyplot as plt
from google.colab import drive
```

```python
#lendo o dataset
df=pd.read_excel('/content/drive/MyDrive/CURSOS/Estudos - Data Analitycs/Tic - Análise de dados com paython /Módulo 3/Aula 1/meteorite_landings.xlsx')
df.head()
```

```python
#Gráfico de linhas
d=df['year'].value_counts()
met=pd.DataFrame({'ano': d.index, 'cont': d.values}).sort_values(by='ano')
met=met[(met['ano']>1958) & (met['ano']<2013)]
plt.plot(met['ano'], met['cont'])
plt.show()
```

```python
#Mudança de cor
plt.xlabel('Ano')
plt.ylabel('Quantidade')
plt.plot(met['ano'], met['cont'], color='red')
plt.show()
```

```python
#gráfico de barras
barras=df.groupby(by='recclass').agg({'mass (g)':'mean','id':'count'}).reset_index()
barras=barras[barras['id']>300]
```

```python
plt.bar(barras['recclass'], barras['mass (g)'])
plt.show()
```

```python
#valores de média em cada barra
_, ax = plt.subplots()
bar_container = ax.bar(barras['recclass'], barras['mass (g)'])
ax.set(ylabel='Média da massa', title='clases')
ax.bar_label(bar_container, fmt='{:.0f}')
plt.show()
```

```python
#Gráfico de barras horizontal
plt.barh(barras['recclass'], barras['mass (g)'])
plt.show()
```

```python
#Mudando a cor
plt.xlabel('Classes')
plt.ylabel('Quantidade')
plt.barh(barras['recclass'], barras['mass (g)'], color='blue')
plt.show()
```

```python
#Gráfico de dispersão
plt.scatter(df['reclong'], df['reclat'])
plt.show()
```

```python
#Removendo os outlier
df.new=df[df['reclong']<300]
plt.scatter(df.new['reclong'], df.new['reclat'])
plt.show()
```

```python
#mudando a cor do gráfico de dispersão
plt.xlabel('Longitude')
plt.ylabel('Latitude')
plt.scatter(df.new['reclong'], df.new['reclat'], color='gray')
plt.show()
```

```python
#salvando figura
plt.xlabel('Longitude')
plt.ylabel('Latitude')
plt.scatter(df.new['reclong'], df.new['reclat'], color='gray')
plt.savefig('/content/drive/MyDrive/CURSOS/Estudos - Data Analitycs/Tic - Análise de dados com paython /Módulo 3/Aula 1/mapa_dos_meteoros.png')
```
