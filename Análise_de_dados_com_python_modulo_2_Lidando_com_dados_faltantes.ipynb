```python
#corrigindo a grafia do titulo da coluna
df_diabetes = df_diabetes.rename(columns={'concentraÃ§ao de glicose': 'concentração de glicose'})
df_diabetes.head()
```

```python
df_diabetes = df_diabetes.rename(columns={'pressao arterial': 'pressão arterial'})
df_diabetes.head()
```

## Descartando linhas com valores faltantes

```python
# por ter poucos dados achei melhor remover. cada coluna represneta menos de 1%
# Remove linhas com valores faltantes em 'concentração de glicose' ou 'imc'
df_diabetes.dropna(subset=['concentração de glicose', 'imc'], inplace=True)
df_diabetes
```

```python
df_diabetes.isnull().sum()
```

```python
df_diabetes.info()
```

## Substituindo valores

```python
df_diabetes.isna().sum()
```

```python
df_diabetes[df_diabetes['insulina'].isna()].head(10)
```

```python
#substituindo o valor para uma informação mais legivel
df_diabetes['leu insulina'] = ~df_diabetes['insulina'].isna()
df_diabetes.head()
```

```python
#retirando a coluna com dados faltantes
df_diabetes.drop(columns='insulina', inplace=True)
df_diabetes.head()
```

```python
#organizando o dataframe
df_diabetes=df_diabetes[[
    'pressão arterial',
    'concentração de glicose',
    'leu insulina', 'tipo sanguineo',
    'idade','imc', 'diabetico',
]]
df_diabetes
```
