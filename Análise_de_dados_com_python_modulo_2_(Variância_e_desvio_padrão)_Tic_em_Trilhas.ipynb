##Compreendendo Variância e Desvio Padrão

```python
import statistics as st
lista1=[0,1,2,3,3,3,4,4,4,5,5,6]
lista2=[0,-1,-2,30,3,3,4,40,4,-50,5,6]

print(st.mean(lista1))
print(st.mean(lista2))
```

```python
st.variance(lista1)
```

```python
st.variance(lista2)
```

```python
sorted(lista2)
```

```python
m = st.mean(lista2)
m
```

```python
dp=st.stdev(lista2)
dp
```

```python
dp2=st.stdev(lista1)
dp2
```

```python
m-dp,m+dp
```

```python
m-2*dp,m+2*dp
```
