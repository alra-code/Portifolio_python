```python
cidades_serie=pd.Series (["São Paulo(SP)",
                            "Manaus(AM)",
                            "Brasília(DF)"])

populacoes_serie=pd.Series ([12.0,
                              6.5,
                              3.0])
```

```python
cidades_populosas=pd.DataFrame({"Cidade": cidades_serie,
                                   "População em milhões": populacoes_serie})
print(cidades_populosas)
```

```python
print(cidades_populosas["Cidade"])
```

```python
print(cidades_populosas.iloc[:2])
```

```python
media_populacao=cidades_populosas["População em milhões"].mean()
print(media_populacao)
print("Média da população:",f'{media_populacao:.2f}',"milhões de habitantes")
```

```python
area_urbana=['914','640','590']
cidades_populosas['Área Urbana (km²)']=area_urbana
print(cidades_populosas)
```

```python
valor_bruto=cidades_populosas['População em milhões']*1_000_000
cidades_populosas['População - Valor Bruto']=valor_bruto
print(cidades_populosas)
cidades_populosas
```

```python
int_area_urbana=cidades_populosas['Área Urbana (km²)'].astype(int)
cidades_populosas['Área Urbana (km²)']=int_area_urbana
cidades_populosas
```

```python
habitantes_por_km2=cidades_populosas['População em milhões']/cidades_populosas['Área Urbana (km²)']
cidades_populosas['Habitantes por km²']=habitantes_por_km2
cidades_populosas
```
