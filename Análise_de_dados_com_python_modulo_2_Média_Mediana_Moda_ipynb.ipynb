##Diferança entre Média, Mediana e Moda

```python
lista=[2,4,6,1,3,0,4,3,5,3,5,4]
sum(lista)/len(lista)
```

```python
st.median(lista) # media
```

```python
lista_sorted=sorted(lista)
lista_sorted
```

```python
s=len(lista_sorted)
n=s//2
if s%2==0:
  print((lista_sorted[n-1]+lista_sorted[n])/2)
else:
  print(lista_sorted[n])
```

```python
st.median(lista) #mediana
```

```python
st.mode(lista) #moda
```
